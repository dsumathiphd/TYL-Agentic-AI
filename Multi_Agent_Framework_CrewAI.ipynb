Multi Agent Framework - CrewAI



```python
# =====================================================
# CREWAI - MULTI-AGENT WORKSHOP PLANNING SYSTEM
# Google Colab | No API Key | No Ollama | No Paid Credits
# Rule-based agents using a custom CrewAI LLM interface
# =====================================================

# Step 1: Install CrewAI

%pip install -q crewai


# Step 2: Import required libraries

from crewai import Agent, Task, Crew, Process, BaseLLM


# Step 3: Define workshop requirements

STUDENTS = 100
MAX_BUDGET = 15000
SPEAKER_COST = 6000
REFRESHMENTS = 5000
MATERIALS = 2000

initial_cost = (
    2 * SPEAKER_COST + REFRESHMENTS + MATERIALS
)

external_speakers = 2
internal_speakers = 0
total_cost = initial_cost

while total_cost > MAX_BUDGET and external_speakers > 0:

    external_speakers -= 1
    internal_speakers += 1

    total_cost = (
        external_speakers * SPEAKER_COST
        + REFRESHMENTS
        + MATERIALS
    )

remaining_budget = MAX_BUDGET - total_cost

status = (
    "APPROVED"
    if total_cost <= MAX_BUDGET
    else "NOT APPROVED"
)


# Step 4: Create a custom rule-based LLM interface
#
# CrewAI normally requires an LLM.
# This custom interface returns deterministic responses
# without connecting to any external AI service.
#
# This is a classroom simulation of agent reasoning,
# not an actual language model.


class WorkshopLLM(BaseLLM):

    def __init__(self, agent_role):

        super().__init__(
            model="local-rule-based-model",
            temperature=0.0
        )

        self.agent_role = agent_role


    def call(
        self,
        messages,
        tools=None,
        callbacks=None,
        available_functions=None,
        **kwargs
    ):

        # Planner Agent response

        if self.agent_role == "planner":

            return f"""
Thought: I have prepared the workshop agenda.

Final Answer:
Workshop: Agentic AI
Duration: One Day
Participants: {STUDENTS}

Session 1: Introduction to Agentic AI
Session 2: Multi-Agent Systems Demonstration
Session 3: Hands-on Implementation

Initial proposal: Two external speakers.
"""


        # Budget Agent response

        elif self.agent_role == "budget":

            return f"""
Thought: I have evaluated the workshop expenses.

Final Answer:
Maximum Budget: Rs.{MAX_BUDGET}
Initial Cost: Rs.{initial_cost}

External Speakers: {external_speakers}
Internal Faculty: {internal_speakers}

Revised Cost: Rs.{total_cost}
Remaining Budget: Rs.{remaining_budget}
Budget Status: {status}
"""


        # Resource Agent response

        elif self.agent_role == "resource":

            return f"""
Thought: I have identified the required resources.

Final Answer:
External Speakers: {external_speakers}
Internal Faculty: {internal_speakers}

Required Facilities:
1. Seminar Hall
2. Projector
3. Computer Laboratory
4. Internet Connection

College facilities have no additional cost.
"""


        # Coordinator Agent response

        elif self.agent_role == "coordinator":

            return f"""
Thought: I have consolidated the workshop plan.

Final Answer:

FINAL WORKSHOP PLAN
----------------------------------------
Workshop          : Agentic AI
Duration          : One Day
Students          : {STUDENTS}

Agenda:
1. Introduction to Agentic AI
2. Multi-Agent Systems Demonstration
3. Hands-on Implementation

External Speakers : {external_speakers}
Internal Faculty  : {internal_speakers}

Resources:
Seminar Hall, Projector, Computer Laboratory

Maximum Budget    : Rs.{MAX_BUDGET}
Initial Cost      : Rs.{initial_cost}
Total Expenditure : Rs.{total_cost}
Remaining Budget  : Rs.{remaining_budget}
Status            : {status}
----------------------------------------
"""

        raise ValueError("Unknown agent role")


    def supports_function_calling(self):

        return False


    def get_context_window_size(self):

        return 8192


# Step 5: Create Planner Agent

planner = Agent(

    role="Workshop Planner",

    goal="Prepare a one-day Agentic AI workshop agenda",

    backstory="You are an experienced academic event planner.",

    llm=WorkshopLLM("planner"),

    allow_delegation=False,

    max_iter=2,

    verbose=True
)


# Step 6: Create Budget Agent

budget = Agent(

    role="Budget Manager",

    goal="Keep workshop expenses within Rs.15000",

    backstory="You manage college event expenses.",

    llm=WorkshopLLM("budget"),

    allow_delegation=False,

    max_iter=2,

    verbose=True
)


# Step 7: Create Resource Agent

resource = Agent(

    role="Resource Manager",

    goal="Identify suitable speakers and workshop facilities",

    backstory="You manage college infrastructure and resources.",

    llm=WorkshopLLM("resource"),

    allow_delegation=False,

    max_iter=2,

    verbose=True
)


# Step 8: Create Coordinator Agent

coordinator = Agent(

    role="Workshop Coordinator",

    goal="Prepare the final workshop plan",

    backstory="You coordinate academic events.",

    llm=WorkshopLLM("coordinator"),

    allow_delegation=False,

    max_iter=2,

    verbose=True
)


# Step 9: Create Planner Task

task1 = Task(

    description="""
    Prepare a one-day Agentic AI workshop
    for 100 students.

    Initially propose two external speakers
    and prepare three technical sessions.
    """,

    expected_output="Workshop agenda",

    agent=planner
)


# Step 10: Create Budget Task

task2 = Task(

    description="""
    Calculate the workshop expenditure.

    External speaker fee: Rs.6000 per speaker
    Refreshments: Rs.5000
    Materials: Rs.2000
    Maximum budget: Rs.15000

    Revise the plan if the budget is exceeded.
    """,

    expected_output="Revised workshop budget",

    agent=budget,

    context=[task1]
)


# Step 11: Create Resource Task

task3 = Task(

    description="""
    Identify suitable speakers and facilities
    based on the revised workshop budget.
    """,

    expected_output="List of required resources",

    agent=resource,

    context=[task1, task2]
)


# Step 12: Create Coordinator Task

task4 = Task(

    description="""
    Combine the workshop agenda, revised budget
    and resource information.

    Prepare the final workshop plan.
    """,

    expected_output="Complete final workshop plan",

    agent=coordinator,

    context=[task1, task2, task3]
)


# Step 13: Create the Crew

crew = Crew(

    agents=[
        planner,
        budget,
        resource,
        coordinator
    ],

    tasks=[
        task1,
        task2,
        task3,
        task4
    ],

    process=Process.sequential,

    memory=False,

    verbose=True
)


# Step 14: Execute the multi-agent system asynchronously

result = await crew.kickoff_async()


# Step 15: Display final workshop plan

print("\n" + "=" * 50)
print("           FINAL WORKSHOP PLAN")
print("=" * 50)

print(result.raw)

print("=" * 50)


# Step 16: Validate the results

assert initial_cost == 19000
assert total_cost == 13000
assert external_speakers == 1
assert internal_speakers == 1
assert remaining_budget == 2000
assert status == "APPROVED"
assert total_cost <= MAX_BUDGET
assert len(result.tasks_output) == 4
assert "13000" in result.raw
assert "2000" in result.raw
assert "APPROVED" in result.raw

print("\nAll validation tests passed successfully!")

# Step 17: Create Mermaid diagram

mermaid_code = """
flowchart TD

    A([Workshop Request])
    B[Planner Agent]
    C[Budget Agent]
    D[Resource Agent]
    E[Coordinator Agent]
    F([Final Workshop Plan])

    T1[Task 1: Prepare Agenda]
    T2[Task 2: Revise Budget]
    T3[Task 3: Allocate Resources]
    T4[Task 4: Consolidate Plan]

    A --> B
    B --> T1
    T1 --> C
    C --> T2
    T2 --> D
    D --> T3
    T3 --> E
    E --> T4
    T4 --> F

    classDef input fill:#DCFCE7,stroke:#16A34A,color:#166534
    classDef agent fill:#DBEAFE,stroke:#2563EB,color:#1E40AF
    classDef task fill:#FEF3C7,stroke:#D97706,color:#92400E
    classDef output fill:#FCE7F3,stroke:#DB2777,color:#9D174D

    class A input
    class B,C,D,E agent
    class T1,T2,T3,T4 task
    class F output
"""

print("\nMERMAID DIAGRAM SOURCE:\n")

print(mermaid_code)


# Step 18: Display Mermaid diagram

import base64
import json
import requests

from IPython.display import Image, display

diagram_data = {

    "code": mermaid_code,

    "mermaid": {"theme": "default"}
}

encoded = base64.urlsafe_b64encode(

    json.dumps(diagram_data).encode("utf-8")

).decode("ascii")

url = "https://mermaid.ink/img/" + encoded

try:

    response = requests.get(url, timeout=60)

    response.raise_for_status()

    if not response.headers.get(
        "Content-Type", ""
    ).startswith("image/"):

        raise ValueError("Invalid image response")

    display(Image(data=response.content))

except Exception as e:

    print("Mermaid rendering unavailable:", str(e))

    print("Copy the Mermaid source into https://mermaid.live/")


# =====================================================
# KEY CONCEPTS:
#
# Agent              : Defines role and responsibility
# Task               : Assigns work to an agent
# Context            : Provides previous task outputs
# Crew               : Combines agents and tasks
# Process.sequential : Executes tasks sequentially
# BaseLLM            : Custom model interface
# kickoff()          : Starts crew execution
#
# This program uses the actual CrewAI framework.
#
# The custom BaseLLM simulates model responses using
# predefined rules. Budget calculations are performed
# before crew execution. Agents do not independently
# reason over previous task outputs in this simulation.
#
# No API key, paid credits, GPU or Ollama is required.
#
# Internet access is required for package installation
# and optional Mermaid image rendering.
#
# Verification:
# Independent end-to-end Colab execution is unverified.
# Assertions validate calculations and crew outputs
# during execution.
# =====================================================
```
